```python
def E_k(mass_kg, velocity_m_per_s):
    """
    Calculate the kinetic energy of an object.

    Parameters:

    mass_kg (float): Mass of the object in kilograms.
    velocity_m_per_s (float): Velocity of the object in meters per second.

    Returns:
    float: Kinetic energy in joules.

    E_k = 1/2 * mass * velocity^2
    """
    return 0.5 * mass_kg * velocity_m_per_s ** 2

#10 metric tons equals 10,000 kilograms
mass_kg = 10000  # mass in kilograms

# 60 km/h -> 60*(1000m/3600s) = 16.6667 m/s
velocity_m_per_s = 60 * (1000 / 3600)  # velocity in meters per second

print(f"Kinetic energy of the object: {E_k(mass_kg, velocity_m_per_s)} joules")
```

```python
def density(mass_kg, side_m, length_m):

    """
        Density is defined as mass per unit volume. 
    """

    # Calculate volume in cubic meters
    vol = side_m ** 2 * length_m

    # Calculate and returning the density in kg/m^3, ensuring we don't divide by zero
    return mass_kg / vol if vol != 0 else 0

# Given values
mass_g = 200
side_cm = 3
length_cm = 15

# Calculate, convert and print the density

print(f"Density of the object: {density(mass_g/1000, side_cm/100, length_cm/100):.2f} kg/m^3")
```

```python
def divisibility(x):
    """
    Check if a number is divisible by 3, 5, or both.
    I'll utilize the modulo operator to determine divisibility.
    """

    if x%2==0 and x % 3 == 0 and x % 5 == 0: # checking if the remainder for all three is zero
        print(f"{x} is divisible by 2, 3, and 5.")
    elif x % 3 == 0 and x % 5 == 0: # checking if the remainder for both 3 and 5 is zero
        print(f"{x} is divisible by 3 and 5.")
    elif x % 2 == 0 and x % 3 == 0: # checking if the remainder for both 2 and 3 is zero
        print(f"{x} is divisible by 2 and 3.")
    elif x % 2 == 0 and x % 5 == 0: # checking if the remainder for both 2 and 5 is zero
        print(f"{x} is divisible by 2 and 5.")
    elif x % 3 == 0: # checking if the remainder for 3 is zero
        print(f"{x} is divisible by 3.")
    elif x % 5 == 0: # checking if the remainder for 5 is zero
        print(f"{x} is divisible by 5.")
    else: # if none of the above conditions are met, it means the number is not divisible by 2, 3, or 5
        print("not divisible by 2, 3 or 5.")


#Testing code for 12, 7 and 60:

divisibility(12)
divisibility(7)
divisibility(60)
```

```python
def power_sum(x, n):
    n = int(n) # Convert n to an integer to ensure it's a whole number for the range function
    sum = 0 # Initialize sum to 0
    for i in range(0, n + 1): # Loop from 0 to n
        sum += x ** i # Add x raised to the power of i to the sum
    return sum # Return the final sum

# Testing the power_sum function  for x = 2.7 and n = 8.

result = power_sum(2.7, 8) # Compute the sum of powers

print(f"The sum of powers for x = 2.7 and n = 8 is: {result:.2f}") # Print the result with 2 decimal places

```

### A dodgy spring

`1. Put your well-commented code to make the plot below. Be sure to run your notebook so the plot appears!`

```python
import matplotlib.pyplot as plt
import numpy as np

# Inputs
x = np.array([0.05, 0.1, 0.15, 0.2])  # Extension in meters
f_measured = np.array([1.42, 2.44, 3.14, 4.6])  # Force in Newtons
k_theoretical = 30  # N/m

# Theoretical forces
f_values = np.multiply(k_theoretical, x)

# Plotting
plt.figure(figsize=(7, 5))
plt.plot(x, f_measured, "o", label="Measured Data")
plt.plot(
    x, f_values, "-", label=rf"Predicted Force ($k = {k_theoretical}\text{{ N/m}}$)"
)

plt.xlabel("Extension (m)") # labeling the x-axis
plt.ylabel("Force (N)") # labeling the y-axis
plt.title("Force vs. Extension for a Spring (Hooke's Law)") #labeling the title of the graph
plt.grid(True) # adding a grid
plt.legend() # displaying the legend
plt.show()

# Data Analysis
k_individual = np.divide(f_measured, x)
mean_k = np.mean(k_individual)
std_k = np.std(k_individual, ddof=1)

print(f"Individual k values (N/m): \n{k_individual}") # printing the individual k values
print(f"Mean k: {mean_k:.2f} N/m") # printing the mean k value
print(f"Standard Deviation: {std_k:.2f} N/m") # printing the standard deviation of k values
```

`2. Do you think the given value for k is accurate? Explain your reasoning.`

I don't think the the spring constant is accurate as on the $F$ vs. $x$ plot, every measured force point lies significantly below the theoretical line predicted by $k = 30\text{ N/m}$.

I have also calculated and printed out the mean and standard deviation to show where the interval of the spring constant (k) should lie.

The mean experimental spring constant calculated from the measurements is 
$\bar{k} \approx 24.18\text{ N/m}$ with a sample standard deviation of $s_k \approx 3.14\text{ N/m}$.

The interval $\bar{k} \pm s_k$ yields:

$$\text{Expected Interval: } 21.04\text{ N/m} \le k \le 27.32\text{ N/m}$$

Since the advertised value of $30\text{ N/m}$ lies outside this experimental interval, the springs are significantly weaker than labelled.
