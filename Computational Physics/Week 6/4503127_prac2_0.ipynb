```python
def toMPerSec(kmPerH):
    """
    Convert speed from kilometers per hour to meters per second.

    Takes in a speed in kilometers per hour 
    
    1 km -> 1000 m
    1 h -> 3600 s
    """
    return kmPerH * 1000 / 3600

print(toMPerSec(42)) 
```

```python
""" 2. Write some python code to calculate the resistance of a resistor with a potential difference of 5V across it
and a 1mA current running through it (hint: remember Ohm’s law?)."""

def R(V, I):
    """
    Calculate the resistance of a resistor using Ohm's law.

    Takes in a potential difference (V) in volts and current (I) in amperes.
    
    Ohm's law: R = V / I
    """
    return V / I

resistance = R(5, 1/1000)  # 1 mA = 0.001 A
rounded_R = f"{resistance:.2f}"  # Round to two decimal places
print(rounded_R, "\u03A9")  # Output: 5000.00 Ohms
```

### Question 3: Conditional statements

```python
def check_x(x):
    """
    Check if x>0, x<0, or x==0.
    """
    if x > 0: #Positive condition
        print("positive")
    elif x < 0: #Negative condition
        print("negative")
    else: #Zero condition
        print("zero")

def check_string_length(s):
    """
    Check the length of a string.

    Takes in a string s and prints whether it is too long or ok.
    """
    if len(s) > 5: #Long string condition
        print("too long")
    else: #If the string is not longer than 5 characters then it is ok
        print("ok")

print("Check x:") #Test for Q3.1
check_x(7)   # Output: positive
check_x(-3)  # Output: negative
check_x(0)   # Output: zero

print("\nCheck string length:") #Separator for Q3.2
check_string_length("cosmological")  # Output: too long
check_string_length("ohm")           # Output: ok
```

### Question 4: Loops

```python
def count_letters(s):
    """
    Count the number of letters in a string using a loop.

    Takes in a string s and returns the count of its letters.
    """
    count = 0
    for char in s:
        count += 1
    return count

def sum_positive_numbers(num_list):
    """
    Sum all positive numbers in a list using a loop.

    Takes in a list of numbers and returns the sum of positive numbers.
    """
    total = 0 # Initialize total to 0, we'll update this as we loop through the list
    for n in num_list:
        if n > 0: #Check if the number is positive
            total += n #add that number to the total if it is
    return total

# Test the functions
print("Count letters:")
print(count_letters("antidisestablishmentarianism"))  # Output: 28
print("\nSum positive numbers:")
print(sum_positive_numbers([-5, -9, 3, -1, 11, 0]))  # Output: 14
print(sum_positive_numbers([2, -1, 9, 0]))  # Output: 11


```

### Question 5: Find x at the maximum

```python
def f(x):
    """
    Calculate the value of the function f(x) = (x^4)/(x-11) + 100x^2.

    Takes in a value x and returns the computed value of the function.
    """
    return (x**4) / (x - 11) + 100 * (x**2)

x=1.0  # Start searching from x = 1.0
max_x = x  # Initialize max_x to the starting value
max_f = f(x)  # Initialize max_f to the function value at the starting point

while x <= 10.0:  # Search until x reaches 10.0

    if f(x) > max_f:  # If the current function value is greater than the max found so far
        max_f = f(x)  # Update max_f to the new maximum value
        max_x = x  # Update max_x to the current x value

    x += 0.1  # Increment x by 0.1 for the next iteration

print(f"The value of x that maximizes f(x) is approximately: {max_x:.1f}")  # Output the result rounded to one decimal place
```
