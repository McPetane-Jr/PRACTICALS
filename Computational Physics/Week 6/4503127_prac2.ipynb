````python
def sum_positive_numbers(num_list):
    """
    Sum all positive numbers in a list using a loop.

    Takes in a list of numbers and returns the sum of positive numbers.
    """
    total = 0 # Initialize total to 0, we'll update this as we loop through the list
    for n in num_list:
        if n > 0: #Check if the number is positive
            total += n #add that number to the total if it is
    return total

print("\nSum positive numbers:")
print(sum_positive_numbers( [-5,-9, 3,-1, 11, 0]))  # prints output
print(sum_positive_numbers([2, -1, 9, 0]))  #prints output
````

## 10.5 Rocket motion — the central idea

Your textbook begins with a simple rocket model: a rocket ejects fuel backwards, and the rocket moves forwards. The key principle is conservation of linear momentum.

### 1. Start with a single ejection of fuel

Imagine a rocket initially at rest. Its mass is \(M\), and it contains a block of fuel of mass \(m\). The fuel is ejected backwards at speed \(u\) relative to the rocket.

After ejection, the rocket moves forwards at speed \(v\). The fuel's velocity relative to the ground is \(v-u\).

Because the system starts at rest and there is no external force affecting its momentum,

$$
\text{Initial momentum}=\text{Final momentum}
$$

So,

$$
0=Mv+m(v-u)
$$

Expand and rearrange:

$$
0=Mv+mv-mu
$$

$$
(M+m)v=mu
$$

Therefore,

$$
\boxed{v=\frac{m}{M+m}u}
$$

What does this tell us? Ejecting fuel backwards gives the rocket forward momentum. A greater exhaust speed \(u\), or a greater fuel mass \(m\), produces a greater recoil speed for the same rocket mass \(M\).

Notice that \(M+m\) is the total mass before ejection. After ejection, the rocket itself has mass \(M\).

### 2. What changes when the rocket burns fuel continuously?

A real rocket ejects small amounts of fuel continuously, so its mass changes with time.

Gregory derives the rocket equation in free space:

$$
\boxed{m\frac{dv}{dt}=(-\dot m)u}
$$

Here:

- \(m=m(t)\) is the rocket's current mass, including its remaining fuel.
- \(v=v(t)\) is its velocity.
- \(u\) is the exhaust speed relative to the rocket.
- \((\dot m)=\frac{dm}{dt}<0\), because the rocket is losing mass.

The quantity \(-\dot m\) is positive, so the thrust is

$$
\boxed{T=(-\dot m)u}
$$

The rocket accelerates because it ejects momentum backwards. This equation assumes free space, with no external force such as gravity.

### 3. Deriving the rocket velocity

If the exhaust speed \(u\) is constant, divide the rocket equation by \(m\):

$$
dv=-u\frac{dm}{m}
$$

Integrate from initial mass \(m_0\) and initial velocity \(v_0\) to current mass \(m\) and velocity \(v\):

$$
\int_{v_0}^{v}dv
=-u\int_{m_0}^{m}\frac{dm}{m}
$$

This gives

$$
v-v_0=-u\ln\left(\frac{m}{m_0}\right)
$$

Therefore, the rocket velocity equation is

$$
\boxed{v=v_0+u\ln\left(\frac{m_0}{m}\right)}
$$

The logarithm appears because the rocket's acceleration depends on its changing mass.
