```python
import numpy as np
import matplotlib.pyplot as plt

'''
The captured data is as follows:

Distance from the center of the bow handle (to the string I think) in cm : mass attached to the bow string



23.2 cm: 0

23.9 cm: 0.5 kg -> this was the mass hanger, I think.

27.2 cm: +1 kg -> first mass piece I think + hanger.

28.7 cm: +2 lb - >2nd mass piece

30.3 cm: +2 lb ->3rd mass piece.

31.3 cm: +1 lb -> 4th

32.3 cm: +1lb -> 5th

33.3 cm: +1 lb -> 6th

cm to m (/100)
lb to kg (*0.453592)
'''

bow_yf = np.array([23.2, 23.9, 27.2, 28.7, 30.3, 31.3, 32.3, 33.3]) / 100

#==============Cumulative masses===============
m0, m1, m2, m3, m4, m5, m6, m7 = 0, 0.5, 1.0 + 0.453592 * 2, 0.453592 * 2, 0.453592 * 1, 0.453592 * 1, 0.453592 * 1, 0.453592 * 1
# for i in range(1, 7):
#     globals()[f'm{i}'] = globals()[f'm{i-1}'] + globals()[f'm{i}']
    # this is a bit of a hack, but it works. It takes the previous mass and adds the current mass to it.
a =m0
b= a+m1
c= b+m2
d= c+m3
e= d+m4
f= e+m5
g= f+m6

masses = np.array([a, b, c, d, e, f, g])

#============== Extension of the bow string = distance from the center of the bow handle - 23.2 cm (the initial position) = bow_yf - 0.232 m
extension = bow_yf - 0.232

plt.plot(extension, masses, 'o-')
plt.xlabel('Extension of the bow string (m)')
plt.ylabel('Mass attached to the bow string (kg)')
plt.title('Mass vs Extension of the Bow String')
plt.grid()
plt.show()


    
```

```python


'''
Before we dive in the rest of the data, here's the setup. We tried shooting the arrow using the same string extension throughout, which is the 26.7 cm - 23.2 ig. Then one of us had that wheel thingie that measures the meters as you push it on the ground, he moved from the point of fire to where the arrow landed, he probably didn't always move straight which is why the repeated trials. We had 4 trials for both the 30 deg and 45 deg angles



There might be human error in the data



Horizontal displacement of the arrow at 30 deg

Distance (m) : Time (s)

43.6: 3.14

47.2: 3.06

46.1: 2.58

49.8: 2.56





Horizontal displacement of the arrow at 45 deg

Distance (m) : Time (s)

55.8: 3.85

55.6: 3.28

50: 3.20

50.8: 2.93
'''

#============30 degr data=========================
x_30_deg = np.array([43.6, 47.2, 46.1, 49.8])
t_30_deg = np.array([3.14, 3.06, 2.58, 2.56])

# Average of the 30 deg data
x_30_mean = np.mean(x_30_deg)
t_30_mean = np.mean(t_30_deg)

# Uncertainty of the 30 deg data
x_30_uncertainty = np.std(x_30_deg, ddof=1) / np.sqrt(len(x_30_deg))
t_30_uncertainty = np.std(t_30_deg, ddof=1) / np.sqrt(len(t_30_deg))

#============45 degr data=========================
x_45_deg = np.array([55.8, 55.6, 50, 50.8])
t_45_deg = np.array([3.85, 3.28, 3.20, 2.93])

# Average of the 45 deg data
x_45_mean = np.mean(x_45_deg)
t_45_mean = np.mean(t_45_deg)

# Uncertainty of the 45 deg data
x_45_uncertainty = np.std(x_45_deg, ddof=1) / np.sqrt(len(x_45_deg))
t_45_uncertainty = np.std(t_45_deg, ddof=1) / np.sqrt(len(t_45_deg))

#===========Printing the results=========================
print("30 deg data:")
print(f"Average distance: {x_30_mean:.2f} m ± {x_30_uncertainty:.2f} m")
print(f"Average time: {t_30_mean:.2f} s ± {t_30_uncertainty:.2f} s")
print("\n45 deg data:")
print(f"Average distance: {x_45_mean:.2f} m ± {x_45_uncertainty:.2f} m")
print(f"Average time: {t_45_mean:.2f} s ± {t_45_uncertainty:.2f} s")
```
